```python
%load_ext autoreload
%autoreload 2

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from pathlib import Path
import sys
sys.path.insert(0, str(Path().resolve().parents[2] / 'src'))
from functions import plot_results_session_summary

sns.set_context('paper', rc={
    'xtick.major.size': 1,
    'ytick.major.size': 1,
})
```

```python
save_data = str(DATA_DIR / 'supp_figures' / 'supp_fig_6_generalization_delay_code') + '/'
save_figures = str(FIGURES_OUT / 'supp_fig_6') + '/'
```

```python
colors = ['black', 'steelblue', 'crimson']
baseline = 0

fig, axes = plt.subplots(3, 2, figsize=(12, 9), sharey=True)

decoder_types = ['stimulusv2', 'delay', 'responsev2']
delay_trains  = [3, 10]

# Epoch shading: (x0, x1, color, alpha)
MARKERS_STIM  = [(0, 0.4, 'grey', 0.4), (3.4, 3.6, 'black', 0.4), (10.4, 10.6, 'steelblue', 0.4)]
MARKERS_GOCUE = [(0, 0.2, 'black', 0.4), (-3, -2.6, 'grey', 0.4), (-10, -9.6, 'steelblue', 0.4)]

panel_labels = ['a', 'b', 'c', 'd', 'e', 'f']
label_idx = 0

for row_idx, decode in enumerate(decoder_types):
    for col_idx, delay_train in enumerate(delay_trains):

        if decode == 'delay':
            file_name = f'/review_crossdelay_decoder_train_{delay_train}.csv'
        else:
            file_name = f'/review_crossdelay_decoder_train_{decode}_{delay_train}.csv'

        df_cum = pd.read_csv(save_data + file_name, index_col=0)

        if decode == 'responsev2':
            align, window = 'Delay_FF', [-12, 4]
            all_cum = pd.DataFrame()
            for value_to_subtract in [3, 10]:
                delay = df_cum.loc[df_cum['delay'] == value_to_subtract].copy()
                delay.columns = [
                    str(float(col) - value_to_subtract)
                    if pd.notna(pd.to_numeric(col, errors='coerce')) else col
                    for col in delay.columns
                ]
                all_cum = pd.concat([all_cum, delay], ignore_index=True)
        else:
            align, window = 'Stimulus_ON', [-2, 14]
            all_cum = df_cum.copy()

        ax      = axes[row_idx, col_idx]
        markers = MARKERS_STIM if align == 'Stimulus_ON' else MARKERS_GOCUE

        plot_results_session_summary(
            None, ax, all_cum, colors,
            variables_combined=['WM_3', 'WM_10'],
            y_range=[baseline - 0.1, baseline + 0.5],
            x_range=window,
            epoch=align,
            baseline=baseline,
            epoch_markers=markers,
        )

        if row_idx == 0:
            ax.set_title(f'Train Delay: {delay_train}s', fontsize=12, fontweight='bold')
        ax.text(-0.10, 1.05, panel_labels[label_idx], transform=ax.transAxes,
                fontsize=14, fontweight='bold', va='top', ha='right')
        label_idx += 1
        if col_idx == 1:
            ax.tick_params(labelleft=True)

plt.tight_layout(w_pad=5, h_pad=2)
# fig.savefig(save_figures + '/review_decoder_across_delays_combined_grid.svg', dpi=300)
plt.show()
```
